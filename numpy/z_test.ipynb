```python
import numpy as np
from scipy.stats import norm

# Simulate a large sample of size 40
sample = np.random.normal(loc=172, scale=3, size=40)  # mean ~172, std dev 3
population_mean = 170
population_std = 3

samp_mean = np.mean(sample)
n = len(sample)

z_score = (samp_mean - population_mean) / (population_std / np.sqrt(n))
print("Sample mean:", samp_mean)
```
